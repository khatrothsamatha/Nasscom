```python
import statistics

scores = [72, 85, 90, 68, 74, 85, 92, 78, 85, 88, 63, 91]

mean   = statistics.mean(scores)
median = statistics.median(scores)
mode   = statistics.mode(scores)
print(f'Data   : {sorted(scores)}')
print(f'Mean   : {mean:.2f}')
print(f'Median : {median}')
print(f'Mode   : {mode}')

```

```python
import statistics
scores = [72, 85, 90, 68, 74, 85, 92, 78, 85, 88, 63, 91]

data_range  = max(scores) - min(scores)
variance    = statistics.variance(scores)
std_dev     = statistics.stdev(scores)

print(f'Range      : {data_range}')
print(f'Variance   : {variance:.2f}')
print(f'Std Dev    : {std_dev:.2f}')
```

```python
data = [2, 4, 4, 4, 5, 5, 7, 9]
n    = len(data)

mean_val = sum(data) / n
print(f'Mean = {mean_val}')

deviations        = [x - mean_val for x in data]
squared_devs      = [d ** 2 for d in deviations]
variance_manual   = sum(squared_devs) / n
std_dev_manual    = variance_manual ** 0.5

print(f'Deviations  : {deviations}')
print(f'Sq.Devs     : {squared_devs}')
print(f'Variance    : {variance_manual:.2f}')
print(f'Std Dev     : {std_dev_manual:.2f}')
```

```python

total_outcomes = 6


p_roll_4 = 1 / total_outcomes
print(f'P(rolling 4)         = {p_roll_4:.4f}')


even_outcomes = 3
p_even = even_outcomes / total_outcomes
print(f'P(rolling even)      = {p_even:.4f}')


p_odd = 1 - p_even
print(f'P(rolling odd)       = {p_odd:.4f}')

p_less_3 = 2 / total_outcomes
print(f'P(rolling less than 3) = {p_less_3:.4f}')

```

```python
import random

random.seed(42)
n_rolls = 10000

count_even = 0

for _ in range(n_rolls):
    roll = random.randint(1, 6)
    if roll % 2 == 0:
        count_even += 1

experimental_probability = count_even / n_rolls
print(f'Rolls        : {n_rolls}')
print(f'Even count   : {count_even}')
print(f'Experimental P(even) = {experimental_probability:.4f}')
print(f'Theoretical  P(even) = 0.5000')
```

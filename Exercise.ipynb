```python
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")
DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}

# --- Task 1: Python Fundamentals, Lists, Tuples, Sets, Dictionaries, Indexing & Slicing ---
print("\n--- Task 1: Data Generation and Basic Structures ---")
# ---------------- Task 1.1 ----------------
raw_sensor_data = []

for i in range(1, 101):
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    timestamp = (
        datetime.datetime.now() -
        datetime.timedelta(minutes=random.randint(0, 1000))
    ).strftime("%Y-%m-%d %H:%M:%S")

    device_id = f"DEV{i:03d}"

    reading = (timestamp, device_id, device_type, location, value)
    raw_sensor_data.append(reading)

print("Total Sensor Readings:", len(raw_sensor_data))
print("First Reading:", raw_sensor_data[0])


# ---------------- Task 1.2 ----------------
temperature_readings = []

for reading in raw_sensor_data:
    if reading[2] == "Temperature_Sensor":
        temp_dict = {
            "timestamp": reading[0],
            "device_id": reading[1],
            "location": reading[3],
            "temperature_celsius": reading[4]
        }
        temperature_readings.append(temp_dict)

print("\nTemperature Readings:", len(temperature_readings))


# ---------------- Task 1.3 ----------------
unique_locations = set()

for reading in raw_sensor_data:
    unique_locations.add(reading[3])

print("\nUnique Locations:")
print(unique_locations)


# ---------------- Task 1.4 ----------------
device_stats = {}

for device in DEVICE_TYPES:
    device_stats[device] = (float("inf"), float("-inf"))

for reading in raw_sensor_data:
    device = reading[2]
    value = reading[4]

    current_min, current_max = device_stats[device]

    if value < current_min:
        current_min = value

    if value > current_max:
        current_max = value

    device_stats[device] = (current_min, current_max)

print("\nDevice Statistics:")
for device, stats in device_stats.items():
    print(device, "-> Min:", stats[0], "Max:", stats[1])


# ---------------- Task 1.5.1 ----------------

sliced_readings_1 = raw_sensor_data[9:20]


sliced_readings_2 = raw_sensor_data[::5]

print("\n10th to 20th Readings:", len(sliced_readings_1))
print("Every 5th Reading:", len(sliced_readings_2))


# ---------------- Task 1.5.2 ----------------
sample_string = "PythonProgrammingIsFun"


substring_1 = sample_string[6:17]

substring_2 = sample_string[5::-1]

print("\nSubstring 1:", substring_1)
print("Substring 2:", substring_2)


# ---------------- Task 1.5.3 ----------------

fifth_reading_device_type = raw_sensor_data[4][2]


device_type_suffix = fifth_reading_device_type[-6:]

print("\n5th Reading Device Type:", fifth_reading_device_type)
print("Suffix:", device_type_suffix)

print("\n--- Task 2: Function Definitions and Usage ---")

def get_readings_by_location(reading_list, target_location):
    result = []
    for i in reading_list:
        if i[3] == target_location:
            result.append(i)
    return result

print("\n--- Output Task 2.1 ---")

lab1_readings = get_readings_by_location(raw_sensor_data, "Lab1")

print("Total Readings in Lab1:", len(lab1_readings))

print("\nFirst 5 Readings from Lab1:")
for reading in lab1_readings[:5]:
    print(reading)

def calculate_average_temperature(temp_data_list):
    if len(temp_data_list) == 0:
        return 0.0

    total = 0

    for reading in temp_data_list:
        total += reading["temperature_celsius"]

    average = total / len(temp_data_list)
    return round(average, 2)


# Example
avg_temp = calculate_average_temperature(temperature_readings)
print("Average Temperature:", avg_temp, "°C")

def generate_report_line(reading_dict):
    timestamp = reading_dict["timestamp"]
    device_id = reading_dict["device_id"]
    location = reading_dict["location"]
    temperature_celsius = reading_dict["temperature_celsius"]

    return f"[{timestamp}] Device {device_id} at {location}: {temperature_celsius}°C"

print("\n--- Output Task 2.3 ---")

if temperature_readings:
    report = generate_report_line(temperature_readings[0])
    print(report)
else:
    print("No temperature readings available.")

# --- Task 3: Working with Core Python Libraries (os, datetime, random) ---
print("\n--- Task 3: Core Library Usage ---")



# Your code for Task 3.1 here:
reports_dir = 'sensor_reports'
if not os.path.exists(reports_dir):
    os.makedirs(reports_dir)
    print(f"\nDirectory '{reports_dir}' created.")
else:
    print(f"\nDirectory '{reports_dir}' already exists.")

print("\n--- Output Task 3.1 ---")
print("Current Working Directory:")
print(os.getcwd())

print("Reports Folder:", reports_dir)

# Your code for Task 3.2 here:
if len(raw_sensor_data) > 1:
    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]


    date_format = '%Y-%m-%d %H:%M:%S'
    first_dt = datetime.datetime.strptime(first_timestamp_str, date_format)
    last_dt = datetime.datetime.strptime(last_timestamp_str, date_format)

    duration = last_dt - first_dt

    days = duration.days
    seconds = duration.total_seconds()
    hours = int(seconds // 3600) % 24
    minutes = int((seconds % 3600) // 60)
    print(f"\nDuration of data collection: {days} days, {hours} hours, {minutes} minutes.")
else:
    print("\nNot enough data to calculate duration.")

print("\n--- Output Task 3.2 ---")

print("First Timestamp:", first_timestamp_str)
print("Last Timestamp :", last_timestamp_str)

print("Duration:", duration)


# Your code for Task 3.3 here:
random_values = []

for i in range(5):
    random_values.append(random.randint(100,200))

print(random_values)

print("\n--- Output Task 3.3 ---")

print("Random Values:")
print(random_values)

print("One value per line:")
for value in random_values:
    print(value)



# --- Task 4: File Handling in Python ---
print("\n--- Task 4: File I/O ---")



# Your code for Task 4.1 here:
report_filename = os.path.join(reports_dir, 'temperature_report.txt')
try:
    with open(report_filename, 'w') as f:
        for reading_dict in temperature_readings:
            line = generate_report_line(reading_dict)
            f.write(line + '\n')
    print(f"\nTemperature report written to '{report_filename}'.")
except IOError as e:
    print(f"\nError writing file '{report_filename}': {e}")

print("\n--- Output Task 4.1 ---")

print("Report File:", report_filename)

print("File Exists:", os.path.exists(report_filename))

print("File Size:", os.path.getsize(report_filename), "bytes")



# Your code for Task 4.2 here:
read_report_lines = []

try:
    with open(report_filename,"r") as f:
        read_report_lines = f.readlines()

    print("First 3 lines:")

    for line in read_report_lines[:3]:
        print(line.strip())

except IOError as e:
    print(e)

print("\n--- Output Task 4.2 ---")

print("Total Lines in Report:", len(read_report_lines))

print("\nComplete Report:")

for line in read_report_lines:
    print(line.strip())


print("\n--- Exercise Complete! ---")
print("Great job completing the intermediate Python exercise.")
print("Review your code and outputs to ensure correctness and understanding of all concepts.")
print("Consider adding more complex scenarios or user inputs for further practice.")

```
